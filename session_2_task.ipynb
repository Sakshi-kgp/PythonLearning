### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
# Write code here
def calculate_in_hand_salary(ctc_lakhs):
    # Convert Lakhs to actual value for calculation
    ctc = ctc_lakhs * 100000

    # 1. Calculate Fixed Deductions (HRA: 10%, DA: 5%, PF: 3%)
    hra = 0.10 * ctc
    da = 0.05 * ctc
    pf = 0.03 * ctc
    total_fixed_deductions = hra + da + pf

    # 2. Calculate Tax Deduction based on CTC brackets
    if ctc_lakhs < 5:
        tax_rate = 0.0
    elif 5 <= ctc_lakhs <= 10:
        tax_rate = 0.10
    elif 10 < ctc_lakhs <= 20:
        tax_rate = 0.20
    else: # Above 20
        tax_rate = 0.30

    tax_deduction = ctc * tax_rate

    # 3. Final Calculation
    yearly_net = ctc - total_fixed_deductions - tax_deduction
    monthly_in_hand = yearly_net / 12

    return round(monthly_in_hand, 2)

# Input
ctc_input = float(input("Enter your yearly CTC in Lakhs: "))
monthly_salary = calculate_in_hand_salary(ctc_input)

print(f"--- Salary Breakdown for {ctc_input} Lakhs ---")
print(f"Monthly In-hand Salary: ₹{monthly_salary:,.2f}")
````

### `Problem 2`: Write a program that take a user input of three angles and will find out whether it can form a triangle or not.

````python
# Write code here
def is_triangle(a, b, c):
    # Check if sum is 180 and all angles are positive
    if (a + b + c == 180) and a > 0 and b > 0 and c > 0:
        return True
    else:
        return False

# Taking user input
try:
    angle1 = float(input("Enter first angle: "))
    angle2 = float(input("Enter second angle: "))
    angle3 = float(input("Enter third angle: "))

    if is_triangle(angle1, angle2, angle3):
        print(f"Yes, the angles {angle1}, {angle2}, and {angle3} can form a triangle.")
    else:
        print(f"No, these angles cannot form a triangle. (Sum: {angle1 + angle2 + angle3} deg)")
except ValueError:
    print("Please enter valid numerical values for the angles.")
````

### `Problem 3`: Write a program that will take user input of cost price and selling price and determines whether its a loss or a profit.

````python
# Write code here
def check_profit_loss():
    try:
        # Taking user input
        cp = float(input("Enter the Cost Price (CP): "))
        sp = float(input("Enter the Selling Price (SP): "))

        if sp > cp:
            profit = sp - cp
            percentage = (profit / cp) * 100
            print(f"Result: Profit of {profit:.2f}")
            print(f"Profit Percentage: {percentage:.2f}%")

        elif cp > sp:
            loss = cp - sp
            percentage = (loss / cp) * 100
            print(f"Result: Loss of {loss:.2f}")
            print(f"Loss Percentage: {percentage:.2f}%")

        else:
            print("Result: No Profit, No Loss (Break-even).")

    except ValueError:
        print("Invalid input! Please enter numerical values.")

# Run the program
check_profit_loss()
````

### `Problem 5` - Exercise 12: Display Fibonacci series up to 10 terms.

*Note: The Fibonacci Sequence is a series of numbers. The next number is found by adding up the two numbers before it. The first two numbers are 0 and 1. For example, 0, 1, 1, 2, 3, 5, 8, 13, 21. The next number in this series above is 13+21 = 34*

````python
# Write code here
# Number of terms required
n_terms = 10

# Initialize the first two terms
n1, n2 = 0, 1
count = 0

print("Fibonacci series up to 10 terms:")

# Loop to generate and print terms
while count < n_terms:
    print(n1, end="  ")
    # Calculate the next term
    nth = n1 + n2
    # Update values for the next iteration
    n1 = n2
    n2 = nth
    count += 1

````

###`Problem 13`:Print all the Armstrong numbers in a given range.
Range will be provided by the user<br>
Armstrong number is a number that is equal to the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

````python
# Write code here
n=int(input("Enter first no"))
m=int(input("Enter second no"))

for i in range(n,m+1):
  y=i
  ans=0
  while y>0:
    m=y%10
    ans=ans+m**3
    y=y//10
  if ans==i:
    print(i)

````

###`Problem 14`:Calculate the angle between the hour hand and minute hand.

Note: There can be two angles between hands; we need to print a minimum of two. Also, we need to print the floor of the final result angle. For example, if the final angle is 10.61, we need to print 10.

Input:<br>
H = 9 , M = 0<br>
Output:<br>
90<br>
Explanation:<br>
The minimum angle between hour and minute
hand when the time is 9 is 90 degress.

````python
import math

def calculate_clock_angle(H, M):
    # Normalize hour to 12-hour format
    H = H % 12

    # Calculate angles relative to 12:00
    hour_angle = (H * 30) + (M * 0.5)
    minute_angle = M * 6

    # Find absolute difference
    angle = abs(hour_angle - minute_angle)

    # Get the smaller of the two possible angles
    min_angle = min(angle, 360 - angle)

    # Return the floor of the final result
    return math.floor(min_angle)

# Example Execution
# H = 9, M = 0 -> Output: 90
print(calculate_clock_angle(9, 0))



````

###`Problem 15`:Given two rectangles, find if the given two rectangles overlap or not. A rectangle is denoted by providing the x and y coordinates of two points: the left top corner and the right bottom corner of the rectangle. Two rectangles sharing a side are considered overlapping. (L1 and R1 are the extreme points of the first rectangle and L2 and R2 are the extreme points of the second rectangle).

Note: It may be assumed that the rectangles are parallel to the coordinate axis.

<img src='https://www.geeksforgeeks.org/wp-content/uploads/rectanglesOverlap.png' width='300' height='200'>

````python
# Write code here
def do_overlap(L1, R1, L2, R2):
    # L1, R1, L2, R2 are lists/tuples: [x, y]

    # 1. If one rectangle is to the left of the other
    if L1[0] > R2[0] or L2[0] > R1[0]:
        return False

    # 2. If one rectangle is above the other
    # (Assuming y increases upwards: Top is L.y, Bottom is R.y)
    if R1[1] > L2[1] or R2[1] > L1[1]:
        return False

    return True

# Example Usage:
# Rect 1: Top-Left (0, 10), Bottom-Right (10, 0)
# Rect 2: Top-Left (5, 5), Bottom-Right (15, -5)
print(do_overlap([0, 10], [10, 0], [5, 5], [15, -5]))  # Output: True
````
